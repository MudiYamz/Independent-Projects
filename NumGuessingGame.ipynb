```python
import random

print("Hi! Welcome to the Number Guessing Game!\nYou have 10 attempts to guess the number. Let's start the game!")

low = int(input("Enter the lower limit of the range: "))
high = int(input("Enter the upper limit of the range: "))

num = random.randint(low,high)
# Total number of attempts allowed
ch = 10
gc = 0

while gc < ch:
    gc += 1
    guess = int(input(f"Guess a number between {low} and {high}: "))

    if guess == num:
        print(f"Congratulations! You've guessed the number {num} in {gc} attempts.")
        break

    if gc >= ch:
        print(f"Sorry, you've used all your attempts. The number was {num}. Better luck next time!")
        break

    if guess < num:
        print("Your guess is too low.")
    elif guess > num:
        print("Your guess is too high.")
```
